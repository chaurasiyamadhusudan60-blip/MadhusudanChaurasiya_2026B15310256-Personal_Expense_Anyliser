## 4. NumPy practice
Convert one or more numeric columns to a NumPy array. Try `np.mean()`, `np.max()` and `np.min()` where appropriate.


```python
import numpy as np

# Check your column names
print("Columns:", df.columns.tolist())

# Automatically select the numeric column (replaces df['column_name'])
numeric_data = df.select_dtypes(include='number').iloc[:, 0]

# Convert to a NumPy array
num_array = numeric_data.to_numpy()

# Calculate average, maximum and minimum
print("Average:", np.mean(num_array))
print("Maximum:", np.max(num_array))
print("Minimum:", np.min(num_array))
```

## 5. Pandas analysis
Create at least two calculated columns or useful summaries. Filter and sort rows to answer questions about your dataset.


```python
# Grab the FIRST numeric column name as a text string
col = df.select_dtypes(include='number').columns[0]

# TODO: Create calculated column(s)
df['Tax_10_Percent'] = df[col] * 0.10
df['Total_With_Tax'] = df[col] + df['Tax_10_Percent']

# TODO: Filter records using a condition
high_expenses = df[df[col] > 200]
print("Expenses > 200:\n", high_expenses)

# TODO: Sort records and display the top/bottom results
sorted_df = df.sort_values(by=col, ascending=False)
print("\nTop 5 highest expenses:\n", sorted_df.head())
print("\nBottom 5 lowest expenses:\n", sorted_df.tail())
```

## 6. Your individual project challenge
**Challenge:** Calculate total expenses; category-wise totals and averages; identify largest expense category.

Write your solution below and explain what the result means.


```python
# TODO: Solve your assigned challenge
# Automatically grab the amount column and category column
amount_col = df.select_dtypes(include='number').columns[0]
category_col = df.select_dtypes(include=['object', 'string']).columns[-1]

# 1. Total expenses
total_expenses = df[amount_col].sum()
print("Total Expenses:", total_expenses)

# 2. Category-wise totals and averages
category_summary = df.groupby(category_col)[amount_col].agg(['sum', 'mean'])
print("\nCategory Summary:\n", category_summary)

# 3. Identify largest expense category
largest_category = category_summary['sum'].idxmax()
largest_amount = category_summary['sum'].max()
print(f"\nLargest Expense Category: {largest_category} ({largest_amount})")
```
